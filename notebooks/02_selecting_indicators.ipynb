# Predicting Madrid Housing Prices Using Socioeconomic Indicators using Machine Learning

```python
# Import the Pandas library for data manipulation and analysis
import pandas as pd

# Import NumPy for numerical operations
import numpy as np

# Import Matplotlib for creating basic charts and graphs
import matplotlib.pyplot as plt

# Import Seaborn for creating attractive statistical visualisations
import seaborn as sns 
```

```python
# Loading Dataset 1: Panel of Sociodemographic Indicators of Districts and Neighborhoods of Madrid
indicators = pd.read_csv("../data/indicators.csv",sep=";")
```

```python
# Create a clean working copy of dataset
indicators_clean = indicators.copy()
```

```python
# Display the first five rows
indicators_clean.head()
```

```python
# Rename columns

indicators_clean.rename(columns={
    "Orden": "row_no",
    "Periodo panel": "dataset_version",
    "ciudad": "city",
    "cod_distrito": "district_code",
    "distrito": "district_name",
    "cod_barrio": "neighbourhood_code",
    "barrio": "neighbourhood_name",
    "año": "year",
    "fecha_indicador": "date",
    "fuente_indicador": "data_source",
    "categoría_1": "main_category",
    "categoría_2": "subcategory",
    "indicador_nivel1": "indicator_level1",
    "indicador_nivel2": "indicator_level2",
    "indicador_nivel3": "indicator_level3",
    "unidad_indicador": "unit",
    "indicador_completo": "indicator_name",
    "valor_indicador": "indicator_value"
}, inplace=True)
```

```python
# Display the first ten rows following column rename
indicators_clean.head(10)
```

```python
# Checking dataset coverage across different years
indicators_clean["dataset_version"].value_counts().sort_index()
```

```python
#Display unique dataset names
indicators_clean["neighbourhood_name"].nunique()
```

```python
# Remove rows where neighbourhood_name is missing
indicators_clean_n= indicators_clean.dropna(subset=["neighbourhood_name"])
```

```python
indicators_clean_n.sample(10)
```

```python
# Obtaining the neighbourhood names to compare for obtaining a foreign key for the other dataset

indicators_clean_n[
    ["neighbourhood_code", "neighbourhood_name"]
].drop_duplicates().head(20)
```

```python
#Display information for 1 neighbourhood in 2025

palacio = indicators_clean_n.loc[
    (indicators_clean_n['dataset_version'] == 2025) &
    (indicators_clean_n['neighbourhood_name']== 'Palacio')]
palacio

# There are 119 rows 
```

```python
# Display the diffrerent main categories of the indicators

palacio["main_category"].value_counts()
```

```python
# Display the diffrerent subcateogries of the indicators

palacio["subcategory"].value_counts()
```

```python
# Understanding if each neighbourhood in 2025 has the same amount of rows or if there are likely whitespaces
# part 1: Create a catalog grouped by neighbourhood name
neighbourhood_catalog = (indicators_clean_n[indicators_clean_n["dataset_version"] == 2025]
    .groupby("neighbourhood_name")
    .size()
    .reset_index(name="row_count")
)
neighbourhood_catalog
```

```python
# Understanding if each neighbourhood in 2025 has the same amount of rows or if there are like whitespaces 
# part 2: Save to Excel
neighbourhood_catalog.to_excel(
    "../outputs/neighbourhood_catalog.xlsx",
    index=False
)
```

```python
# Retain only columns with information on types of indicies using palacio 
n_indicator_catalog= palacio[
    [
        "main_category",
        "subcategory",
        "indicator_name",
        "unit"
    ]]
```

```python
# Remove duplicates
n_indicator_catalog = n_indicator_catalog.drop_duplicates()
```

```python
# Sort alphabetically
n_indicator_catalog = n_indicator_catalog.sort_values(
    ["main_category", "subcategory", "indicator_name"]
)
```

```python
# Save to Excel
n_indicator_catalog.to_excel(
    "../outputs/n_indicator_catalog.xlsx",
    index=False
)
```

# Predicting Average House Prices in Madrid Neighbourhoods Using Socioeconomic Indicators

## Overall Project Aim:
To develop and evaluate a machine learning model that uses 2025 socioeconomic indicators to predict average housing prices across Madrid neighbourhoods.

## 2. Feature Selection

### Introduction
Following the initial exploration of the raw data, the socioeconomic indicators included in the dataset will need to be examined to identify appropriate features for the machine learning model. The purpose of this notebook is to explore these indicators and select relevant features for subsequent analysis and modelling.

### Import Libraries

```python
# Import the Pandas library for data manipulation and analysis
import pandas as pd
```

### Load Dataset

```python
# Loading Dataset 1: Panel of Sociodemographic Indicators of Districts and Neighborhoods of Madrid
indicators = pd.read_csv("../data/indicators.csv",sep=";")
```

```python
# Create a clean working copy of dataset
indicators_clean = indicators.copy()
```

### Initial Inspection and Column Rename

```python
# Display the first five rows
indicators_clean.head()
```

```python
# Rename columns

indicators_clean.rename(columns={
    "Orden": "row_no",
    "Periodo panel": "dataset_version",
    "ciudad": "city",
    "cod_distrito": "district_code",
    "distrito": "district_name",
    "cod_barrio": "neighbourhood_code",
    "barrio": "neighbourhood_name",
    "año": "year",
    "fecha_indicador": "date",
    "fuente_indicador": "data_source",
    "categoría_1": "main_category",
    "categoría_2": "subcategory",
    "indicador_nivel1": "indicator_level1",
    "indicador_nivel2": "indicator_level2",
    "indicador_nivel3": "indicator_level3",
    "unidad_indicador": "unit",
    "indicador_completo": "indicator_name",
    "valor_indicador": "indicator_value"
}, inplace=True)
```

```python
# Display the first ten rows following column rename
indicators_clean.head(10)
```

```python
# Checking dataset coverage across different years
indicators_clean["dataset_version"].value_counts().sort_index()
```

### Observations 
As the model aims to use the most recent available data, 2025 was selected as the year of analysis. A check was conducted to confirm that 2025 had comparable data coverage to other recorded years, helping to validate its suitability for use in the model.

### Observations

Therefore, it is was decided to standardise the neighbourhood names in the housing prices dataset to match the format used in this dataset, creating a consistent key for merging.

### Observations

Exploration of the 2025 data for the Palacio neighbourhood shows 119 rows of information, covering 11 main categories and 18 subcategories. While these categories can provide useful context for organising the data, observations are heavily concentrated within a small number of categories. Therefore, feature selection will focus primarily on the individual indicator_name values, which provide more specific information about the socioeconomic characteristics available for modelling.

```python
# Check the number of indicator records per neighbourhood in 2025
neighbourhood_catalog = (indicators_clean_n[indicators_clean_n["dataset_version"] == 2025]
    .groupby("neighbourhood_name")
    .size()
    .reset_index(name="row_count")
)
neighbourhood_catalog
```

```python
# Saving the number of indicator records per neighbourhood in 2025 in an excel
neighbourhood_catalog.to_excel(
    "../outputs/neighbourhood_catalog.xlsx",
    index=False
)
```

### Observations 

The 131 neighbourhoods contain between 106 and 122 indicator records for 2025. While the number of records varies, overall indicator coverage appears broadly similar across neighbourhoods.

```python
# Retain only columns with information on types of indicies using palacio 
n_indicator_catalog= palacio[
    [
        "main_category",
        "subcategory",
        "indicator_name",
        "unit"
    ]]
```

```python
# Remove duplicates
n_indicator_catalog = n_indicator_catalog.drop_duplicates()
```

```python
# Sort alphabetically
n_indicator_catalog = n_indicator_catalog.sort_values(
    ["main_category", "subcategory", "indicator_name"]
)
```

```python
# Save to Excel
n_indicator_catalog.to_excel(
    "../outputs/n_indicator_catalog.xlsx",
    index=False
)
```

### Observations

To simplify exploration of the large dataset, indicator information for Palacio was selected as a representative neighbourhood. The main_category, subcategory, indicator_name and unit columns were exported to a catalogue for further review and feature selection.

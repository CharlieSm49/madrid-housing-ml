# Predicting Average House Prices in Madrid Neighbourhoods Using Socioeconomic Indicators

## Overall Project Aim:
To develop and evaluate a machine learning model that uses 2025 socioeconomic indicators to predict average housing prices across Madrid neighbourhoods.

## 3. Data Cleaning and Preparation

### Introduction

Based on the initial exploration of the dataset, the following data cleaning and preprocessing steps were carried out on a working copy of the dataset:

- Renamed columns to provide descriptive English labels.
- Converted relevant columns to appropriate numerical data types and reformat values to standard decimal notation.
- Removed city and district-level data, retained only neighbourhood-level observations.
- Filtered the dataset to include only 2025 data.
- Restructured the data so that each neighbourhood represents one row and each feature represents one column.
- Retained only the indicators required for the 15 selected features. Where a feature required calculation from multiple indicators, the necessary component indicators were temporarily retained, combined to create the feature, and then removed.

### Format values to standard decimal notation and into a numerical data type

### Obtaining the 15 socioeconomic indicators

```python
# Create the two new features

# Percentage of registered housing over total registered premises *
pivoted_features['percentage_residential_properties']= pivoted_features['registered_residential_properties']/pivoted_features['total_registered_properties']*100

# Percentage of population aged 25+ with higher education (university degree, postgraduate or doctorate) or diploma/technical degree *
pivoted_features['percentage_of_population_25_plus_either_higher_ed_or_technical']= (pivoted_features['population_25_plus_higher_education']+ pivoted_features['population_25_plus_technical_diploma'])*100/ (pivoted_features['population_25_plus_lower_secondary_education']+ pivoted_features['population_25_plus_incomplete_primary_education']+pivoted_features['population_25_plus_higher_education']+pivoted_features['population_25_plus_upper_secondary_education']+pivoted_features['population_25_plus_unknown_education_level']+pivoted_features['population_25_plus_technical_diploma']+pivoted_features['population_25_plus_no_formal_education'])
```

```python
# Confirm column names and datatypes
pivoted_features.dtypes
```

```python
# Drop columns that are no longer needed
features_2025_prepared= pivoted_features.drop(columns=["population_25_plus_lower_secondary_education",
                                                          "population_25_plus_incomplete_primary_education",
                                                          "population_25_plus_higher_education",
                                                          "population_25_plus_upper_secondary_education",
                                                          "population_25_plus_unknown_education_level",
                                                          "population_25_plus_technical_diploma",
                                                          "population_25_plus_no_formal_education",
                                                          "registered_residential_properties",
                                                          "total_registered_properties"])
                                                          
                                                          
```

```python
# Remove the label for columns left from pivot

features_2025_prepared.columns.name = None
```

Aeropuerto (Airport) was also removed as it is not a residential neighbourhood, improving alignment with the model's aim of predicting residential housing prices across Madrid neighbourhoods and with the preparation of the housing price dataset.

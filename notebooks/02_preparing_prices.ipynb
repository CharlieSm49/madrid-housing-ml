# Predicting Average House Prices in Madrid Neighbourhoods Using Socioeconomic Indicators

## Import Libraries

```python
# Import the Pandas library for data manipulation and analysis
import pandas as pd
```

## Load Datasets

```python
# Loading Dataset 2: Average Declared Housing Price (euros/m2) by District and Neighborhood in Madrid
prices = pd.read_csv("../data/prices.csv", sep=";",skiprows=5)
```

## 2. Data Cleaning

### Observations 
Based on the initial exploration of the dataset, the following data cleaning and preprocessing steps will be carried out on a working copy of the dataset:

- Combine the header and the first two rows to create meaningful column names.
- Remove the final four rows containing metadata.
- Rename columns to provide descriptive English labels.
- Convert relevant columns to the appropriate numerical data types and reformat values to standard decimal notation.
- Remove rows representing the City of Madrid and district-level data, retaining only neighbourhood-level observations.
- Filter the dataset to include only 2025 data.

In addition, the neighbourhood names will be examined and standardised where necessary to facilitate merging with the socioeconomic indicators dataset. This will ensure that a common key is available for joining the two datasets.

```python
# Create a working copy of dataset
prices_clean = prices.copy()
```

```python
# Display the first ten rows
prices_clean.head(10)
```

#### Combine the header and the first two rows to create appropriate column names

```python
# Combine first two rows into header
new_columns = ( prices_clean.columns.astype(str) + "_" +
                prices_clean.iloc[1].fillna("").astype(str)+ ":_" +
                prices_clean.iloc[0].fillna("").astype(str) )

prices_clean.columns= new_columns
```

```python
# Drop first two rows by resetting the indexes
prices_clean = prices_clean.iloc[2:].reset_index(drop=True)
```

```python
# Display the first ten rows
prices_clean.head(10)
```

#### Filter the dataset to include only 2025 data

```python
# Inspect the last 9 columns containing 2025 data

prices_clean.columns[-9:]
```

```python
# Create a new DF with 2025 prices data
prices_clean_2025_unformatted= prices_clean[["Unnamed: 0_:_","Unnamed: 1_:_",'2025_Total:_Euros/m2']].copy()
```

```python
prices_clean_2025_unformatted.head(10)
```

#### Rename columns to provide descriptive English labels.

```python
# Rename columns 

column_mapper = {'Unnamed: 0_:_': 'district_name', 
                'Unnamed: 1_:_': 'neighbourhood_name',
                '2025_Total:_Euros/m2': 'mean_average_price_euros_per_m2',}

av_prices_clean_2025=prices_clean_2025_unformatted.rename(mapper=column_mapper, axis=1)
```

```python
# Display DF 
av_prices_clean_2025.head()
```

#### Remove the final four rows containing metadata and Remove rows representing the city of Madrid

```python
# Count missing values
av_prices_clean_2025.isnull().sum()
```

```python
# Display rows with missing values 
av_prices_clean_2025[av_prices_clean_2025.isnull().any(axis=1)]
```

```python
# Drop rows with missing values and 1st row as that is a city wide metric  
av_prices_clean_2025 = av_prices_clean_2025.drop([0,153,154,155,156])
```

```python
# Display dimensions
av_prices_clean_2025.shape
```

### Observations 

Following the removal of the metadata and the first row representing the City of Madrid, the dataset contains three columns representing the district name, neighbourhood name and average housing price. The dataset also contains 152 rows.

Madrid is divided into 21 official districts and 131 neighbourhoods. As 21 + 131 = 152, this confirms that the dataset now contains one observation for each district and neighbourhood, with only 2025 housing price data remaining.

```python
# Display last 30 rows

av_prices_clean_2025.tail(30)
```

### Remove rows representing district-level data, retaining only neighbourhood-level data

```python
# Display rows where district matches neighbourhood being distrect level observations
district_prices_2025= av_prices_clean_2025[
    av_prices_clean_2025["district_name"] == av_prices_clean_2025["neighbourhood_name"]
]

district_prices_2025
```

```python
# Display dimensions on districts

district_prices_2025.shape
```

```python
# Remove districts averages 

av_prices_clean_2025 = av_prices_clean_2025.loc[av_prices_clean_2025["district_name"] != av_prices_clean_2025["neighbourhood_name"]]
```

```python
# Display dimensions

av_prices_clean_2025.shape
```

#### Observations

Inspection of the rows where the district name and neighbourhood name are identical shows that these represent district-level observations rather than neighbourhood-level data. As the project focuses on neighbourhood-level analysis, these rows were removed from the dataset. This was confirmed using the shape attribute.

```python
# Display info on the DF
av_prices_clean_2025.info()
```

```python
# Previwing the first 5 rows

av_prices_clean_2025.head()
```

```python
#### Create a common key for merge the two datasets
```

```python
# Return the number of distinct district names
av_prices_clean_2025['neighbourhood_name'].nunique()
```

```python
# Obtain the distinct neighbourhood names in order to compare to the neighbourhood names in the  prices dataset
av_prices_clean_2025["neighbourhood_name"].unique()
```

```python
# Format neighbourhood names to remove the number, the dot and space.
av_prices_clean_2025.loc[:, "neighbourhood_name"] = (
    av_prices_clean_2025["neighbourhood_name"]
    .str.replace(r"^\d+\.\s*", "", regex=True)
    .str.strip()
)
```

```python
# View first 5 rows to confirm

av_prices_clean_2025.head()
```

### Format values to standard decimal notation

```python
# Format the values: remove the full stops and replace commas with decimal point

av_prices_clean_2025 = av_prices_clean_2025.replace(r"\.", "", regex=True)
av_prices_clean_2025= av_prices_clean_2025.replace(",", ".", regex=True)
```

```python
# Create exportable DF only of neighbourhood and price value
average_price_2025_prepared= av_prices_clean_2025[["neighbourhood_name","mean_average_price_euros_per_m2"]]
```

```python
# View first 10 rows

average_price_2025_prepared.head(10)
```

#### Convert values to a numerical data type

```python
# Convert all values to float 

average_price_2025_prepared['mean_average_price_euros_per_m2'].astype(float)
```

```python
# Confirm data types as float
average_price_2025_prepared.dtypes
```

```python
# Sort data alphabetically by neighbourhood name and reset index 

average_price_2025_prepared=average_price_2025_prepared.sort_values('neighbourhood_name')

average_price_2025_prepared=average_price_2025_prepared.reset_index(drop=True)
```

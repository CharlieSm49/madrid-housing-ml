```python
# Create a list of feature names

selected_features = [
    "Población densidad (hab./ha.)",
    "Población mayor/igual de 25 años con bachiller elemental, graduado escolar, ESO, formación profesional 1º grado",
    "Población mayor/igual de 25 años con enseñanza primaria incompleta",
    "Población mayor/igual de 25 años con estudios superiores, licenciatura, arquitectura, ingeniería sup., estudios sup. no universitarios, doctorado, postgraduado",
    "Población mayor/igual de 25 años con formación profesional 2º grado, bachiller superior o BUP",
    "Población mayor/igual de 25 años con nivel de estudios desconocido y/o no consta",
    "Población mayor/igual de 25 años con titulación media, diplomatura, arquitectura o ingeniería técnica",
    "Población mayor/igual de 25 años que no sabe leer ni escribir o sin estudios",
    "Número de locales dados de alta en uso como vivienda",
    "Número de locales dados de alta totales",
    "Renta media disponible por persona",
    "Renta neta media anual de los hogares (Urban Audit)",
    "Tasa absoluta de paro registrado (febrero)",
    "Tasa de crecimiento demográfico (porcentaje)",
    "Tamaño medio del hogar",
    "Edad media de la población",
    "Porcentaje de envejecimiento (población mayor de 65 años/población total)",
    "Índice de vulnerabilidad bienestar social e igualdad",
    "Índice de vulnerabilidad economía y empleo",
    "Índice de vulnerabilidad educación y cultura",
    "Índice de vulnerabilidad medio ambiente urbano y movilidad",
    "Índice de vulnerabilidad salud"
]
```

```python
# Pivot table so that each row is one neighbourhood
pivoted_features=filtered_features_2025.pivot(index="neighbourhood_name",columns="indicator_name",values="indicator_value")
```

```python
# Rename columns 

pivoted_features.rename(columns={

    "Edad media de la población":
        "average_population_age",

    "Número de locales dados de alta en uso como vivienda":
        "registered_residential_properties",

    "Número de locales dados de alta totales":
        "total_registered_properties",

    "Población densidad (hab./ha.)":
        "population_density",

    "Población mayor/igual de 25 años con bachiller elemental, graduado escolar, ESO, formación profesional 1º grado":
        "population_25_plus_lower_secondary_education",

    "Población mayor/igual de 25 años con enseñanza primaria incompleta":
        "population_25_plus_incomplete_primary_education",

    "Población mayor/igual de 25 años con estudios superiores, licenciatura, arquitectura, ingeniería sup., estudios sup. no universitarios, doctorado, postgraduado":
        "population_25_plus_higher_education",

    "Población mayor/igual de 25 años con formación profesional 2º grado, bachiller superior o BUP":
        "population_25_plus_upper_secondary_education",

    "Población mayor/igual de 25 años con nivel de estudios desconocido y/o no consta":
        "population_25_plus_unknown_education_level",

    "Población mayor/igual de 25 años con titulación media, diplomatura, arquitectura o ingeniería técnica":
        "population_25_plus_technical_diploma",

    "Población mayor/igual de 25 años que no sabe leer ni escribir o sin estudios":
        "population_25_plus_no_formal_education",

    "Porcentaje de envejecimiento (población mayor de 65 años/población total)":
        "ageing_rate",

    "Renta media disponible por persona":
        "average_disposable_income_per_person",

    "Renta neta media anual de los hogares (Urban Audit)":
        "average_annual_net_household_income",

    "Tamaño medio del hogar":
        "average_household_size",

    "Tasa absoluta de paro registrado (febrero)":
        "registered_unemployment_rate",

    "Tasa de crecimiento demográfico (porcentaje)":
        "population_growth_rate",

    "Índice de vulnerabilidad bienestar social e igualdad":
        "social_welfare_and_equality_vulnerability_index",

    "Índice de vulnerabilidad economía y empleo":
        "economy_and_employment_vulnerability_index",

    "Índice de vulnerabilidad educación y cultura":
        "education_and_culture_vulnerability_index",

    "Índice de vulnerabilidad medio ambiente urbano y movilidad":
        "urban_environment_and_mobility_vulnerability_index",

    "Índice de vulnerabilidad salud":
        "health_vulnerability_index"

}, inplace=True)
```

```python
# View first 5 rows
pivoted_features.head()
```

```python
# Check columns

pivoted_features.columns
```

```python
# Sort data alphabetically by neighbourhood name

pivoted_features=pivoted_features.sort_values('neighbourhood_name')
```

```python
# Format the values: remove the full stops and replace commas with decimal point

pivoted_features = pivoted_features.replace(r"\.", "", regex=True)
pivoted_features = pivoted_features.replace(",", ".", regex=True)
```

```python
# Change values to float 

pivoted_features=pivoted_features.astype(float)
```

```python
# View first 5 rows
pivoted_features.head()
```

```python
# Check columns

pivoted_features.columns
```

```python
# Reset index 
pivoted_features = pivoted_features.reset_index()
```

```python
# Confirm data types as float and check columns 
pivoted_features.dtypes
```

```python
# Create new features

pivoted_features['percentage_residential_properties']= pivoted_features['registered_residential_properties']/pivoted_features['total_registered_properties']*100

pivoted_features['percentage_of_population_25_plus_either_higher_ed_or_technical']= (pivoted_features['population_25_plus_higher_education']+ pivoted_features['population_25_plus_technical_diploma'])*100/ (pivoted_features['population_25_plus_lower_secondary_education']+ pivoted_features['population_25_plus_incomplete_primary_education']+pivoted_features['population_25_plus_higher_education']+pivoted_features['population_25_plus_upper_secondary_education']+pivoted_features['population_25_plus_unknown_education_level']+pivoted_features['population_25_plus_technical_diploma']+pivoted_features['population_25_plus_no_formal_education'])
```

```python
# Confirm column names and datatypes
pivoted_features.dtypes
```

```python
# Drop columns that are no longer needed
features_2025_prepared= pivoted_features.drop(columns=["population_25_plus_lower_secondary_education",
                                                          "population_25_plus_incomplete_primary_education",
                                                          "population_25_plus_higher_education",
                                                          "population_25_plus_upper_secondary_education",
                                                          "population_25_plus_unknown_education_level",
                                                          "population_25_plus_technical_diploma",
                                                          "population_25_plus_no_formal_education",
                                                          "registered_residential_properties",
                                                          "total_registered_properties"])
                                                          
                                                          
```

```python
# Remove the label for columns left from pivot

features_2025_prepared.columns.name = None
```

```python
# Confirm column names and datatypes
features_2025_prepared.dtypes
```
